program 10:Write a Python program that takes five marks, stores them in a list, and calculates the average.

```python
#average=total/len(list)
def calculate_average():#code runs line by line
  marks=[]
  for i in range(5):
    numbers=float(input("enter 5 numbers "))
    marks.append(numbers)

  total=0
  for numbers in marks:
    total+=numbers
  average=total/len(marks)
  print(average)

calculate_average()
```
